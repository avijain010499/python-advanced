## all()
```text
- all() returns True only if every element in the iterable is True
- If even one element is False → returns False

`Use Case 1: Multi-Rule Validation`
```text
Real-world use:
- Business rule validation
- Schema validation
- Data quality enforcement

````python
age = 25
salary = 50000

if all([age > 18, salary > 0]):
    print("Valid record")
````

`Use Case 2: Check for Empty or Invalid Fields`
```text
Used in:
- Checking missing data
- Ensuring mandatory fields exist

````python
record = {"name": "John", "age": 30,}

is_complete = all(record.values())
print(is_complete)
````

## any()
```text
- any() returns True if at least one element in the iterable is True.
- If all elements are False → returns False.

`Use Case 1: Validation Check`
```text
Real-world use:
- Checking if any validation rule failed
- Detecting presence of bad records
- Monitoring system alerts

````python
errors = [False, False, True]

if any(errors):
    print("Error detected")
````

`Use Case 2: Condition with Generator Expression`
```text
Explanation:
- It checks if any number is greater than 25.

> Important:
This uses generator expression → memory efficient.

````python
nums = [10, 20, 30]

result = any(x > 25 for x in nums)
print(result)
````

### Important Concept: Short-Circuiting

`Both any() and all() stop early`
```text
For any():
- Stops at first True.

For all():
- Stops at first False.
