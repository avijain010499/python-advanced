## enumerate() --> *iterator of index-value pairs*
```text
Enumerate is a built-in function that adds a counter to an iterable and returns an enumerate object, which is an iterator of index-value pairs

- It is mainly used to avoid manual index management.

> Works with: List, String, Tuples, Generators, Files


`Syntax:`
```python
enumerate(iterable, start=0)
```
---
```python
- iterable → list, tuple, string, etc.
- start → optional starting index (default = 0)
- Returns → enumerate object (iterator)


```text
> How enumerate Works Internally

Explanation:
- It pairs each element with its index
- Internally, it keeps a counter and increments it after each iteration

````python
names = ["A", "B", "C"]

result = list(enumerate(names))
print(result)
````

`Use Case 1: Row Numbering (Very Common in ETL)`
```text
Why useful:
- Generating row IDs
- Reporting
- Audit logs
- Adding surrogate keys during transformation

````python
records = ["row1", "row2", "row3"]

for row_number, record in enumerate(records, start=1):
    print(row_number, record)
````

`Use Case 2: Modifying List Using Index`
```text
Explanation:
- enumerate provides both index and value, allowing in-place modification

Used when:
- Applying transformation with index dependency
- Updating lists efficiently

````python
nums = [10, 20, 30]

for index, value in enumerate(nums):
    nums[index] = value * 2

print(nums)
````

`Use Case 3: Conditional Index Detection`
```text
Useful for:
- Searching with position tracking
- Error reporting
- Logging exact failure row


````python
nums = [5, 10, 15, 20]

for index, value in enumerate(nums):
    if value == 15:
        print("Found at index:", index)
````

`Use Case 4: File`

````python
with open("file.txt") as f:
    for line_number, line in enumerate(f, start=1):
        print(line_number, line)
````
