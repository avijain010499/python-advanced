### map() --> *iterator*
`map(function, iterable)`

```text
- Applies a function to every element.

Whys???
- Returns an iterator
- Memory efficient
- Useful in transformations

> function → applied to each element
> iterable → list, tuple, etc.
> Returns → iterator (lazy evaluation)

````python
nums = [1, 2, 3]

result = list(map(lambda x: x * 2, nums))
print(result)   # [2, 4, 6]
````

```text
Execution flow:
1 → 2
2 → 4
3 → 6
4 → 8

Each element is passed individually into the function.

> Important: It does not execute immediately. It executes when iterated (lazy evaluation).

`Use Case 1: Basic Transformation`
```text
In this example, map applies a transformation to every element of the list. Each element is passed into the lambda function, which adds 5. So the output becomes [15, 25, 35].

> Real-world analogy:
If I need to increase all prices by a fixed tax percentage before loading into warehouse, this pattern applies.


````python
nums = [10, 20, 30]

result = list(map(lambda x: x + 5, nums))
print(result)
````

`Use Case 2: Using Named Function`
```text
In production systems, instead of lambda, we often use named functions for readability and reusability. Here, each string is passed to the clean_value function, which removes extra spaces and converts it to lowercase.

> Real-world use:
Cleaning raw CSV string data before storing in structured tables.


````python
def clean_value(x):
    return x.strip().lower()

values = ["  Apple ", " Banana "]

result = list(map(clean_value, values))
print(result)
````

`3. Multiple Iterables Use Case`
```text
Map can take multiple iterables. In this case, elements from both lists are passed together into the function. So first 1 and 4 are passed, then 2 and 5, and so on

> Important behavior:
Stops at shortest iterable

> Real-world example:
Merging two numeric columns from two different sources before loading

````python
#Adding columns one at a time
nums1 = [1, 2, 3]
nums2 = [4, 5, 6]

result = list(map(lambda x, y: x + y, nums1, nums2))
print(result)
````

`4. Type Conversion (Very Common in ETL)`
```text
Here, map applies the int function to every string in the list, converting them into integers

> Real-world use:
Converting string timestamps, numeric fields, boolean flags from API responses.

````python
values = ["1", "2", "3"]

result = list(map(int, values))
print(result)
````

`5. Extracting Values from Dictionaries (JSON Processing)`
```text
In this example, map extracts the ‘age’ value from each dictionary. Each record is passed into the lambda function, and the age field is returned

> Real-world example:
Extracting only required columns before loading into data warehouse

````python
records = [
    {"name": "A", "age": 20},
    {"name": "B", "age": 25}
]

ages = list(map(lambda x: x["age"], records))
print(ages)
````
