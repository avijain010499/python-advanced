## zip --> *iterator*

```text
- zip() is a built-in function that aggregates elements from multiple iterables positionally and returns an iterator of tuples
- It pairs elements based on index
```
`Syntax`
```python
zip(iterable1, iterable2, ..., iterableN)

`Use Case 1: Creating Dictionary from Two Lists`
```text
Very common in ETL

Real-world usage:
- Mapping CSV rows to schema
- API response restructuring
- Column alignment before DB insert

````python
columns = ["id", "name", "salary"]
values = [101, "John", 50000]

record = dict(zip(columns, values))
print(record)
````

`Use Case 2: Parallel Iteration (Cleaner Code)`
```text
Instead of manual indexing, zip is used

````python
names = ["A", "B", "C"]
scores = [90, 95, 85]

for name, score in zip(names, scores):
    print(name, score)
````

`Use Case 3: Multiple Iterables`
```text
Used for:
- Multi-column aggregation
- Combining multiple data streams

````python
a = [1, 2, 3]
b = [4, 5, 6]
c = [7, 8, 9]

print(list(zip(a, b, c)))
````

`Use Case 4: Unzipping Data (Very Important)`
```text
The * operator unpacks the list of tuples, effectively transposing rows into columns

Used in:
- Data transformation
- Separating structured fields
- Pivot-like operations

````python
pairs = [(1, "A"), (2, "B"), (3, "C")]

nums, letters = zip(*pairs)

print(nums)
print(letters)
````
