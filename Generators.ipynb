## GENERATORS

A generator is a function that:
- Produces one value at a time
- Pauses after producing a value
- Resumes from the same place next time
- Does not store all values in memory

Key word: **lazy execution**

````python
'''
Ends the function immediately
State is destroyed
'''
def f():
    return 1
    return 2   # never runs

'''
Returns a value
Pauses the function
Saves state
Resumes on next call
'''
# Generator
def gen():
    yield 1
    yield 2

g = gen() # No code runs yet, since 'g' is a generator object
print(next(g)) # Execution starts here --> Prints 1
print(next(g)) # Prints 2
````

### Generator execution step-by-step (VERY IMPORTANT)

````python
def counter(n):
    current = 0
    while current < n:
        current += 1
        yield current

g = counter(3)

print(next(g))
print(next(g))
print(next(g))
print(next(g))
````


#### Timeline:
`g = counter(3)   → nothing runs`

`next(g)          → current=1 → yield 1 → PAUSE` 

`next(g)          → current=2 → yield 2 → PAUSE`

`next(g)          → current=3 → yield 3 → PAUSE`

`next(g)          → StopIteration`


#### Note:
A generator:
- Automatically implements `__iter__()`

- Automatically implements `__next__()`

So:

- Every generator is an iterator
- Not every iterator is a generator


### 1. Why generators are memory efficient?

`nums = (i for i in range(1_000_000))`

- Produces values one by one
- Almost constant memory

### 2. Generator expressions


`squares = (x * x for x in range(5))`

is equivalent to

```python
def squares():
    for x in range(5):
        yield x * x
```

### 3. Generator exhaustion

``` python
g = (i for i in range(3))

list(g)  # [0, 1, 2]
list(g)  # []
```
Once exhausted → cannot be reused





````python
def squares():
    for x in range(5):
       yield x * x
        
gen = squares()
print(list(squares())) # 1. Way to print in list format

# 2. Way to print the values one by one
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
````
