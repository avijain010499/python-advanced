### filter() --> *iterator*
```text
- Filters elements based on condition
- Keeps elements where condition is True
```
`Lazy Evaluation (Very Important Concept)`
```text
Execution happens only when:
- Converted to list
- Iterated in loop

````python
nums = [1, 2, 3, 4]

result = list(filter(lambda x: x % 2 == 0, nums))
print(result)   # [2, 4]
````

`Use Case 1: Basic Conditional Filtering`
```text
Here, filter keeps only numbers greater than 20.

In ETL:
Used before loading to warehouse to discard bad data.

````python
nums = [10, 25, 30, 5]

result = list(filter(lambda x: x > 20, nums))
print(result)
````

`Use Case 2: Removing Null or Empty Values`
```text
filter removes falsy values like empty string and None

Why important:
- Cleaning raw ingestion data
- Removing missing values before transformation
- Preparing clean datasets

````python
values = ["A", "", None, "B", None]

result = list(filter(lambda x: x, values))
print(result)
````

`Use Case 3: Filtering Dictionary Records (JSON Processing)`
```text
Each dictionary is passed to the lambda, and only records where age >= 18 are retained

Real-world use:
- Filtering API response data
- Applying business rules
- Data validation before loading

````python
records = [
    {"name": "A", "age": 20},
    {"name": "B", "age": 15},
    {"name": "C", "age": 25}
]

adults = list(filter(lambda x: x["age"] >= 18, records))
print(adults)
````

`Use Case 4: Using Named Function (Production Style)`
```text
In production systems, we use named functions for clarity and reusability

Why important:
- Improves readability
- Easy to test
- Maintains clean code architecture

````python
def is_valid_salary(x):
    return x > 0

salaries = [50000, -1000, 30000]

result = list(filter(is_valid_salary, salaries))
print(result)
````

`Use Case 5: Filtering Based on Multiple Conditions`
```text
Filter can handle complex conditions. Here, we keep numbers greater than 20 and even

Real-world scenario:
- Keep transactions above threshold and valid status
- Apply multiple validation rules

````python
nums = [10, 25, 40, 5, 60]

result = list(filter(lambda x: x > 20 and x % 2 == 0, nums))
print(result)
````
