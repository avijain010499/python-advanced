## ITERATORS

An iterator is an object that:
- Remembers where it is
- Gives one value at a time
- Stops when data is finished

To be an iterator in Python, an object must implement two methods:
- `__iter__()` → returns the iterator object
- `__next__()` → returns the next value

When values finish → raise **StopIteration**

```python
nums = [10, 20, 30]

it = iter(nums)      # creates iterator
print(next(it))      # 10
print(next(it))      # 20
print(next(it))      # 30
print(next(it))      # StopIteration
```

### Iterable vs Iterator (IMPORTANT)

Iterable -> Can produce an iterator --> Ex: List

Iterator -> Produces values one by one


```python
nums = [1, 2, 3]

iterator = iter(nums) # iterator
iterable = nums         # iterable
for i in iterator:
    print(i)
print(iterable)
```

### Full Iterator Class

#### Important
- `'self'` : refers to the current object being created 

- `'max_value'` : is an argument passed by the user

*Here, our object is **'counter'** i.e 'counter.max_value'*

```python
'''
- current keeps state
- This is why iterators remember position
'''
class Counter:
    def __init__(self, max_value): # 'self' refers to the current object being created, 'max_value' is an argument passed by the user
        self.max_value = max_value # Here, our object is 'counter' i.e 'counter.max_value'
        self.current = 0

    '''
    - Returns the iterator object
    - Required by Python
    '''
    def __iter__(self):
        return self

    '''
    - Returns next value
    - Stops iteration properly
    '''
    def __next__(self):
        if self.current < self.max_value:
            self.current += 1
            return self.current
        raise StopIteration

# Using the iterator
counter = Counter(3)

# StopIteration
print(next(counter))  # 1 
print(next(counter))  # 2
print(next(counter))  # 3
```
