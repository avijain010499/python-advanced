## Functional Programming 

```text
Functional programming is a programming style where:
	•	Functions are treated as first-class objects
	•	We avoid changing state (immutability)
	•	We use pure functions (no side effects)
	•	Functions can be passed as arguments or returned

### 1. First-Class Functions

```text
In Python, functions can:
	•	Be assigned to variables
	•	Be passed as arguments
	•	Be returned from other functions

````python
def square(x):
    return x * x

f = square        # function assigned to variable
print(f(5))       # 25
````

### 2. Lambda Functions

```text
- Anonymous single-line functions.
- Used inside map, filter, sorting, etc.

````python
square = lambda x: x * x
print(square(4))
````

### 3. map() --> *iterator*
`map(function, iterable)`

```text
- Applies a function to every element.

Whys???
- Returns an iterator
- Memory efficient
- Useful in transformations

> function → applied to each element
> iterable → list, tuple, etc.
> Returns → iterator (lazy evaluation)

````python
nums = [1, 2, 3]

result = list(map(lambda x: x * 2, nums))
print(result)   # [2, 4, 6]
````

```text
Execution flow:
1 → 2
2 → 4
3 → 6
4 → 8

Each element is passed individually into the function.

> Important: It does not execute immediately. It executes when iterated (lazy evaluation).

`Use Case 1: Basic Transformation`
```text
In this example, map applies a transformation to every element of the list. Each element is passed into the lambda function, which adds 5. So the output becomes [15, 25, 35].

> Real-world analogy:
If I need to increase all prices by a fixed tax percentage before loading into warehouse, this pattern applies.


````python
nums = [10, 20, 30]

result = list(map(lambda x: x + 5, nums))
print(result)
````

`Use Case 2: Using Named Function`
```text
In production systems, instead of lambda, we often use named functions for readability and reusability. Here, each string is passed to the clean_value function, which removes extra spaces and converts it to lowercase.

> Real-world use:
Cleaning raw CSV string data before storing in structured tables.


````python
def clean_value(x):
    return x.strip().lower()

values = ["  Apple ", " Banana "]

result = list(map(clean_value, values))
print(result)
````

`3. Multiple Iterables Use Case`
```text
Map can take multiple iterables. In this case, elements from both lists are passed together into the function. So first 1 and 4 are passed, then 2 and 5, and so on

> Important behavior:
Stops at shortest iterable

> Real-world example:
Merging two numeric columns from two different sources before loading

````python
#Adding columns one at a time
nums1 = [1, 2, 3]
nums2 = [4, 5, 6]

result = list(map(lambda x, y: x + y, nums1, nums2))
print(result)
````

`4. Type Conversion (Very Common in ETL)`
```text
Here, map applies the int function to every string in the list, converting them into integers

> Real-world use:
Converting string timestamps, numeric fields, boolean flags from API responses.

````python
values = ["1", "2", "3"]

result = list(map(int, values))
print(result)
````

`5. Extracting Values from Dictionaries (JSON Processing)`
```text
In this example, map extracts the ‘age’ value from each dictionary. Each record is passed into the lambda function, and the age field is returned

> Real-world example:
Extracting only required columns before loading into data warehouse

````python
records = [
    {"name": "A", "age": 20},
    {"name": "B", "age": 25}
]

ages = list(map(lambda x: x["age"], records))
print(ages)
````

### 4. filter() --> *iterator*
```text
- Filters elements based on condition
- Keeps elements where condition is True
```
`Lazy Evaluation (Very Important Concept)`
```text
Execution happens only when:
- Converted to list
- Iterated in loop

````python
nums = [1, 2, 3, 4]

result = list(filter(lambda x: x % 2 == 0, nums))
print(result)   # [2, 4]
````

`Use Case 1: Basic Conditional Filtering`
```text
Here, filter keeps only numbers greater than 20.

In ETL:
Used before loading to warehouse to discard bad data.

````python
nums = [10, 25, 30, 5]

result = list(filter(lambda x: x > 20, nums))
print(result)
````

`Use Case 2: Removing Null or Empty Values`
```text
filter removes falsy values like empty string and None

Why important:
- Cleaning raw ingestion data
- Removing missing values before transformation
- Preparing clean datasets

````python
values = ["A", "", None, "B", None]

result = list(filter(lambda x: x, values))
print(result)
````

`Use Case 3: Filtering Dictionary Records (JSON Processing)`
```text
Each dictionary is passed to the lambda, and only records where age >= 18 are retained

Real-world use:
- Filtering API response data
- Applying business rules
- Data validation before loading

````python
records = [
    {"name": "A", "age": 20},
    {"name": "B", "age": 15},
    {"name": "C", "age": 25}
]

adults = list(filter(lambda x: x["age"] >= 18, records))
print(adults)
````

`Use Case 4: Using Named Function (Production Style)`
```text
In production systems, we use named functions for clarity and reusability

Why important:
- Improves readability
- Easy to test
- Maintains clean code architecture

````python
def is_valid_salary(x):
    return x > 0

salaries = [50000, -1000, 30000]

result = list(filter(is_valid_salary, salaries))
print(result)
````

`Use Case 5: Filtering Based on Multiple Conditions`
```text
Filter can handle complex conditions. Here, we keep numbers greater than 20 and even

Real-world scenario:
- Keep transactions above threshold and valid status
- Apply multiple validation rules

````python
nums = [10, 25, 40, 5, 60]

result = list(filter(lambda x: x > 20 and x % 2 == 0, nums))
print(result)
````

### 5. reduce()

`Syntax:` 
```python
from functools import reduce
reduce(function, iterable, initializer)
```
---
```text
Reduces sequence to a single value

- Performs cumulative aggregation
- Useful for summation, multiplication, custom aggregation logic

How It Works
- Takes first two values
- Applies function
- Result is combined with next value
- Continues until one value remains

````python
from functools import reduce

nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)
````

`Use Case 1 : Finding Maximum`
```text
Explanation:
At each step, it keeps the larger value.

````python
from functools import reduce

nums = [5, 10, 3, 8, 99]

maximum = reduce(lambda x, y: x if x > y else y, nums)
print(maximum)
````

`Use Case 2: Flattening a List`
```text
Used in:
- Nested JSON transformations
- Data normalization

````python
from functools import reduce

lists = [[1, 2], [3, 4], [5]]

flat = reduce(lambda x, y: x + y, lists)
print(flat)
````

`Use Case 3: Dictionary Aggregation`
```text
Suppose you want to merge multiple dictionaries:

Useful in:
- Config merging
- Data consolidation

````python
from functools import reduce

dicts = [
    {"a": 1},
    {"b": 2},
    {"c": 3}
]

merged = reduce(lambda x, y: {**x, **y}, dicts)
print(merged)
````

`Use Case 4: Using Initializer`
```text
If initializer is given as 10, then the function will start from 10 
- Initializer acts as starting value.

Execution:
10 + 1 = 11
11 + 2 = 13
13 + 3 = 16

````python
from functools import reduce

nums = [1, 2, 3]

result = reduce(lambda x, y: x + y, nums, 10)
print(result)
````
